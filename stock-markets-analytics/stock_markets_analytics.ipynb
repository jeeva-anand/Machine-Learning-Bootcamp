```python
import yfinance as yf
import pandas as pd

indexes = {
    "United States": "^GSPC",
    "China": "000001.SS",
    "Hong Kong": "^HSI",
    "Australia": "^AXJO",
    "India": "^NSEI",
    "Canada": "^GSPTSE",
    "Germany": "^GDAXI",
    "United Kingdom": "^FTSE",
    "Japan": "^N225",
    "Mexico": "^MXX",
    "Brazil": "^BVSP"
}

data = yf.download(
    list(indexes.values()),
    start="2026-01-01",
    end="2026-08-22",
    auto_adjust=False
)

data.head()
```

```python
close = data["Close"]

ytd_returns = (close.iloc[-1] / close.iloc[0] - 1) * 100
```
